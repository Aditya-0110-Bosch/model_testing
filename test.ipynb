```python
import os
from spire.presentation.common import *
from spire.presentation import *

def ppt_to_images(ppt_file, output_folder="output_images"):
    # Create output folder if it doesn't exist
    if not os.path.exists(output_folder):
        os.makedirs(output_folder)

    # Load the presentation
    presentation = Presentation()
    presentation.LoadFromFile(ppt_file)

    # Iterate through each slide
    for i, slide in enumerate(presentation.Slides):
        # Convert slide to an image stream (default resolution)
        image = slide.SaveAsImage()
        
        # Define output path (e.g., slide_0.png, slide_1.png)
        output_path = os.path.join(output_folder, f"slide_{i}.png")
        
        # Save image and clean up
        image.Save(output_path)
        image.Dispose()

    presentation.Dispose()
    print(f"Done! Images saved in '{output_folder}'")

# Usage
ppt_to_images(r"C:\Users\DQO3KOR\Downloads\sarthi.pptx")
```

```python
import os
from spire.doc import Document, ImageType

def convert_doc_to_images(file_path, output_folder="output_pages"):
    # Ensure output directory exists
    if not os.path.exists(output_folder):
        os.makedirs(output_folder)

    # Load the document
    doc = Document()
    doc.LoadFromFile(file_path)

    # Iterate through each page of the document
    # Note: doc.PageCount is used to determine the range
    for i in range(doc.PageCount):
        # Convert page index 'i' to a bitmap image stream
        image_stream = doc.SaveImageToStreams(i, ImageType.Bitmap)
        
        # Save the stream to a file
        output_filename = os.path.join(output_folder, f"page_{i+1}.png")
        with open(output_filename, 'wb') as f:
            f.write(image_stream.ToArray())
            
    doc.Close()
    print(f"Success! {doc.PageCount} pages saved to '{output_folder}'.")

# Usage
convert_doc_to_images(r"C:\Users\DQO3KOR\Downloads\AI Agents - FABV-Details (1).docx")
```

```python
from docx2pdf import convert
from pdf2image import convert_from_path
import os

def doc_to_images_free(file_path, output_folder="clean_pages"):
    if not os.path.exists(output_folder):
        os.makedirs(output_folder)

    # 1. Convert Docx to temporary PDF (High fidelity, no watermark)
    temp_pdf = "temp_render.pdf"
    convert(file_path, temp_pdf)

    # 2. Convert PDF pages to Images
    images = convert_from_path(temp_pdf)

    for i, image in enumerate(images):
        image.save(os.path.join(output_folder, f"page_{i+1}.png"), "PNG")

    # Cleanup
    os.remove(temp_pdf)
    print(f"Success! {len(images)} clean pages saved.")

doc_to_images_free(r"C:\Users\DQO3KOR\Downloads\AI Agents - FABV-Details (1).docx")
```
